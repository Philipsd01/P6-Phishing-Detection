```python
import matplotlib.pyplot as plt
import numpy as np

# Example data from your logs.
# For each model, we calculate the average of class 0 and class 1 measures.
models = [
    {
        "model": "distilbert-base-uncased",
        "precision": (0.9962 + 0.97896) / 2,  # 0.98758
        "recall": (0.9877 + 0.99343) / 2,       # 0.990565
        "f1": (0.9919 + 0.98615) / 2,           # 0.989025
        "accuracy": 0.98980,
        "fpr": 0.01229
    },
    {
        "model": "roberta-base",
        "precision": (0.9922 + 0.99339) / 2,    # 0.992795
        "recall": (0.9962 + 0.98645) / 2,         # 0.991325
        "f1": (0.9942 + 0.98991) / 2,             # 0.992055
        "accuracy": 0.99265,
        "fpr": 0.00378
    },
    {
        "model": "xlnet-base-cased",
        "precision": (0.9936 + 0.99422) / 2,      # 0.99391
        "recall": (0.9967 + 0.98892) / 2,         # 0.99281
        "f1": (0.9952 + 0.99156) / 2,             # 0.99338
        "accuracy": 0.99385,
        "fpr": 0.00331
    },
    {
        "model": "bert-base-uncased",
        "precision": (0.9929 + 0.98972) / 2,      # 0.99131
        "recall": (0.9941 + 0.98768) / 2,         # 0.99089
        "f1": (0.9935 + 0.98870) / 2,             # 0.99110
        "accuracy": 0.99175,
        "fpr": 0.00591
    }
]

# Extract metrics for each measure across models:
def get_metric_values(metric):
    return np.array([m[metric] for m in models])

metrics_names = ["precision", "recall", "f1", "accuracy"]

# Prepare aggregated summary for graph (averages, min, max)
aggregated = {}
for metric in metrics_names:
    values = get_metric_values(metric)
    avg = np.mean(values)
    min_val = np.min(values)
    max_val = np.max(values)
    aggregated[metric] = {"avg": avg, "min": min_val, "max": max_val}

# For False Positive Rate (fpr)
fpr_values = get_metric_values("fpr")
fpr_avg = np.mean(fpr_values)
fpr_min = np.min(fpr_values)
fpr_max = np.max(fpr_values)

# Plot Graph 1: Precision, Recall, F1-score and Accuracy.
metrics_labels = ["Precision", "Recall", "F1-score", "Accuracy"]
x = np.arange(len(metrics_labels))
avg_vals = [aggregated[m]["avg"] for m in metrics_names]
lower_err = [aggregated[m]["avg"] - aggregated[m]["min"] for m in metrics_names]
upper_err = [aggregated[m]["max"] - aggregated[m]["avg"] for m in metrics_names]
yerr = [lower_err, upper_err]

plt.figure(figsize=(10, 6))
bar1 = plt.bar(x, avg_vals, color="skyblue", edgecolor="black", yerr=yerr, capsize=8)
plt.xticks(x, metrics_labels)
plt.ylabel("Score")
plt.title("Average and Min/Max Values for Precision, Recall, F1-score and Accuracy")
plt.ylim([min(avg_vals) - 0.01, max(avg_vals) + 0.01])
for i, v in enumerate(avg_vals):
    plt.text(x[i] - 0.1, v + 0.001, f"{v:.4f}")
plt.tight_layout()

# Plot Graph 2: False Positive Rate.
plt.figure(figsize=(6, 6))
x2 = np.array([0])
plt.bar(x2, fpr_avg, color="salmon", edgecolor="black", yerr=[[fpr_avg - fpr_min], [fpr_max - fpr_avg]], capsize=8)
plt.xticks(x2, ["False Positive Rate"])
plt.ylabel("FPR")
plt.title("Average and Min/Max for False Positive Rate")
plt.ylim([0, fpr_max + 0.005])
plt.text(x2[0] - 0.1, fpr_avg + 0.0005, f"{fpr_avg:.4f}")
plt.tight_layout()

plt.show()
```
