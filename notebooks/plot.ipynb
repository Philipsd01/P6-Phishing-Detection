```python
# New cell: Grouped Bar Chart with Min/Max error bars for evaluation metrics

import numpy as np

# Define the order of metrics and models you want to display
metrics = ["Accuracy", "Precision", "Recall", "F1"]
models = ["bert", "xlnet", "distilbert", "roberta"]

# Aggregate the data by metric and model
df_agg = df_long.groupby(["metric", "model"])["score"].agg(["mean", "min", "max"]).reset_index()
df_agg.rename(columns={"mean": "mean_score", "min": "min_score", "max": "max_score"}, inplace=True)
df_agg["err_low"] = df_agg["mean_score"] - df_agg["min_score"]
df_agg["err_high"] = df_agg["max_score"] - df_agg["mean_score"]

# Create grouped bar chart
x = np.arange(len(metrics))
width = 0.2  # width for each bar

plt.figure(figsize=(12, 6))
for i, model in enumerate(models):
    # Select and sort the model's data to match the metric order
    model_data = df_agg[df_agg["model"] == model].set_index("metric").loc[metrics].reset_index()
    positions = x + i*width - (width * (len(models)-1)/2)  # center the groups
    plt.bar(positions, model_data["mean_score"], width=width, label=model)
    plt.errorbar(positions, model_data["mean_score"],
                 yerr=[model_data["err_low"], model_data["err_high"]],
                 fmt='none', capsize=5, color='black')

plt.xticks(x, metrics)
plt.title("Model Comparison Across Evaluation Metrics with Range")
plt.ylabel("Score")
plt.xlabel("Metric")
plt.ylim(0.970, 0.986)
plt.legend(title="Model")
plt.tight_layout()
plt.show()
```
